```python
import pickle
import os, sys
import numpy as np
import pandas as pd
from pathlib import Path
from time import process_time
from collections import defaultdict

from prod_fs import ProD
```

```python
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=100, n_features=20, n_classes=5, weights=[0.1, 0.2, 0.3, 0.25, 0.15], n_informative=5, random_state=0
)
y
```

```python
mapping = {0: 'a', 1: 'b', 2: 'c', 3: 'd', 4: 'e'}
y = np.array([mapping[y] for y in y])
y
```

```python
ya = len([i for i in y if i == 'a'])
yb = len([i for i in y if i == 'b'])
yc = len([i for i in y if i == 'c'])
yd = len([i for i in y if i == 'd'])
ye = len([i for i in y if i == 'e'])
print(f"a: {ya}")
print(f"b: {yb}")
print(f"c: {yc}")
print(f"d: {yd}")
print(f"e: {ye}")
```

```python
# Proposed algorithm
tProD_start = process_time()
prodRankerWeighted = ProD(
    integration_method="trapz", delta=500, bw_method="scott",
    k=3, n_jobs=-1, mode="release", lower_end=-1.5, upper_end=2.5,
    averaging_method="weighted"
)
prodRankerWeighted.fit(X, y)
tProD_stop = process_time()
tProDWeighted = tProD_stop - tProD_start
```
